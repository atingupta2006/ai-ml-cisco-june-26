## 2. Category buckets (storage · compute · visualization · ML/MLOps)

```python
buckets = {
    "storage": ["DVC (data versions)", "CSV files (team_sales, Kaggle samples)"],
    "compute": ["Python", "Pandas", "NumPy", "scikit-learn"],
    "visualization": ["matplotlib", "seaborn", "Excel charts"],
    "ML/MLOps": ["scikit-learn", "SHAP", "FeatureTools", "MLflow", "DVC", "FastAPI"],
}

for category, tools in buckets.items():
    print(f"{category}: {', '.join(tools)}")
```

## 6. Checkpoint self-check

```python
mlops_tools = set(reference_matrix.loc[reference_matrix["category"] == "ML/MLOps", "tool"])
assert {"MLflow", "DVC"}.issubset(mlops_tools)

filled_categories = sum(1 for tools in buckets.values() if len(tools) >= 2)
assert filled_categories >= 3
assert len(my_matrix) >= 4
assert org_tool_today.strip() != ""
print("✓ Tool landscape checkpoint passed")
```
