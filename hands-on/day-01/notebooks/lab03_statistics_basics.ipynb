## Descriptive statistics

| Measure | Plain language | Sensitive to outliers? |
|---------|----------------|----------------------|
| **Mean** | Average of all values | Yes |
| **Median** | Middle value when sorted | No |
| **Std dev** | Typical spread around the mean | Yes |
| **Min / Max** | Smallest / largest value | — |

Excel: `=AVERAGE()`, `=MEDIAN()`, `=STDEV.S()` — sample std matches pandas `ddof=1`.

---

## 1. Load team sales data

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

GH_ROOT = Path.cwd().resolve()
if GH_ROOT.name == "notebooks":
    GH_ROOT = GH_ROOT.parents[2]
elif GH_ROOT.name == "day-01":
    GH_ROOT = GH_ROOT.parents[1]
else:
    for parent in [GH_ROOT, *GH_ROOT.parents]:
        if (parent / "hands-on" / "day-01" / "data" / "team_sales.csv").is_file():
            GH_ROOT = parent
            break

TEAM_SALES_CSV = GH_ROOT / "hands-on" / "day-01" / "data" / "team_sales.csv"
df = pd.read_csv(TEAM_SALES_CSV)

print(f"rows: {len(df)}")
display(df.head())
```

---

## 2. Descriptive statistics on `q2_sales`

```python
sales = df["q2_sales"]

stats = {
    "mean": sales.mean(),
    "median": sales.median(),
    "std": sales.std(ddof=1),
    "min": sales.min(),
    "max": sales.max(),
}

print(f"mean q2_sales: {stats['mean']:.2f}")
print(f"median q2_sales: {stats['median']:.2f}")
print(f"std q2_sales: {stats['std']:.2f}")
print(f"min / max q2_sales: {stats['min']} / {stats['max']}")
```

---

## 3. Outlier impact — Team_13

```python
top_team = df.loc[df["q2_sales"].idxmax(), ["team", "region", "q2_sales"]]
print(f"Highest Q2: {top_team['team']} ({top_team['region']}) = {top_team['q2_sales']}")

without_outlier = sales.drop(top_team.name)
print(f"\nMean with Team_13:    {sales.mean():.2f}")
print(f"Mean without Team_13: {without_outlier.mean():.2f}")
print(f"Median (unchanged):   {sales.median():.2f}")
```

The **mean** shifts more than the **median** when one team has unusually high sales.

---

## 4. Visual check — Q2 sales distribution

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(sales, bins=8, edgecolor="white", color="#049fd9")
ax.axvline(stats["mean"], color="#c41230", linestyle="--", label=f"mean = {stats['mean']:.2f}")
ax.axvline(stats["median"], color="#2d8c3c", linestyle=":", label=f"median = {stats['median']:.2f}")
ax.set_xlabel("q2_sales")
ax.set_ylabel("team count")
ax.set_title("Q2 sales distribution (20 teams)")
ax.legend()
plt.tight_layout()
plt.show()
```

---

## 5. Checkpoint

```python
assert len(df) == 20
assert abs(stats["mean"] - 150.30) < 0.01
assert abs(stats["median"] - 148.50) < 0.01
assert abs(stats["std"] - 34.74) < 0.01
assert stats["min"] == 104
assert stats["max"] == 210
assert top_team["team"] == "Team_13"
print("✓ Statistics checkpoint passed")
```
