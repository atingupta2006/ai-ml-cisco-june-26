## 1. Lists — ordered and mutable

A **list** stores items in order. You can append, change, and remove elements after creation.

We will track cities where our demo restaurant chain operates.


```python
cities = ["Bengaluru", "Mumbai", "Delhi", "Hyderabad"]
print("Type:", type(cities))
print("Length:", len(cities))
print("First city:", cities[0])
print("Last city:", cities[-1])
```

### Indexing and slicing

Python uses **zero-based** indexing. Slicing `[start:stop]` returns a new list (shallow copy of that segment).


```python
print("Slice [1:3]:", cities[1:3])  # Mumbai, Delhi

# Mutability: lists can be changed in place
cities.append("Chennai")
print("After append:", cities)
print("Length after append:", len(cities))
```

## 2. Tuples — ordered and immutable

A **tuple** looks like a list but **cannot** be changed after creation. Use tuples for fixed records.

Each city has a tuple of sample customer ratings `(r1, r2, r3, r4)`.


```python
ratings = (3.8, 4.2, 3.5, 4.0)
print("Type:", type(ratings))
print("Tuple:", ratings)
print("Mean rating (manual):", sum(ratings) / len(ratings))
```

### Why immutability helps

- Tuples can be dictionary keys (lists cannot).
- They signal "this record should not change accidentally."
- Functions often return multiple values as tuples: `train_test_split` returns `(X_train, X_test, y_train, y_test)`.

Try uncommenting the line below — you should see a `TypeError`:


```python
# ratings[0] = 5.0  # TypeError: 'tuple' object does not support item assignment
print("Tuple unchanged:", ratings)
```

## 3. Dictionaries — key/value records

A **dict** maps unique **keys** to **values**. JSON from APIs and FastAPI request bodies are dict-like.

We model one restaurant as a dictionary.


```python
restaurant = {
    "name": "Demo Bistro",
    "city": cities[0],       # link to list
    "rating": ratings[0],    # link to tuple
    "cuisines": {"North Indian", "Chinese"},  # set inside dict
}

print("Restaurant record:")
for key, value in restaurant.items():
    print(f"  {key}: {value}")
```

### Adding fields dynamically

Dicts are **mutable**. New keys can appear at runtime — common when enriching a record during ETL.


```python
restaurant["online_order"] = True
restaurant["cuisines"].add("Cafe")  # mutating the nested set

print("Updated restaurant:")
print(restaurant)
```

---

## 4. Sets — unique unordered collections

A **set** stores **unique** items with no guaranteed order. Ideal for distinct cuisines, tags, or user IDs.

Duplicate inserts are silently ignored.


```python
cuisines = restaurant["cuisines"]
print("Set type:", type(cuisines))
print("Unique cuisines:", cuisines)
print("Sorted display:", sorted(cuisines))
print("Count:", len(cuisines))
```

### Set operations (preview)

Later in analytics you may use `&` (intersection) and `|` (union) for tag overlap. Example:


```python
menu_tags = {"spicy", "vegan", "North Indian"}
overlap = cuisines & menu_tags
print("Tags matching menu filter:", overlap)
```

---

## 5. Mutable vs immutable — summary

| Type | Ordered? | Mutable? | Example |
|------|----------|----------|---------|
| `list` | Yes | **Yes** | `cities.append("Pune")` |
| `tuple` | Yes | **No** | `ratings = (3.8, 4.2)` |
| `dict` | Yes (3.7+) | **Yes** | `restaurant["city"] = "Mumbai"` |
| `set` | No | **Yes** | `cuisines.add("Cafe")` |

**Memory tip:** Choose the narrowest type that fits — immutability prevents accidental bugs in long notebooks.
